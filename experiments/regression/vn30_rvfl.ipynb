```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

from loaders._load_vn30_meta import _process_file, VN30, TARGETS
from sklearn.linear_model import RidgeCV, MultiTaskLassoCV
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import r2_score, mean_absolute_percentage_error, root_mean_squared_error

from loaders._load_vn30_reg import preprocess
from models.regression.rvfl import *
```

```python
tracks = {"r2": [], "mape": []}
for symbol in VN30:
    pack = preprocess(symbol, lag=2, val=0.1)

    model = RVFLEnsemble(
        n_nodes=5,
        n_layers=1,
        n_estimators=1,
        activation="tanh",   # có thể thử "relu", "gelu", "sin"...
        n_splits_cv=3,
        base_seed=42,
    )
    r2, mape = model.fit_predict_with_pack(pack)

    tracks["r2"].append(r2)
    tracks["mape"].append(mape)

    print(f"Symbol: {symbol}, R2: {r2:.4f}, MAPE: {mape:.4f}")

print(f"Mean R2 Score: {np.mean(tracks['r2'])}, Mean MAPE: {np.mean(tracks['mape'])}")
print(f"Std R2 Score: {np.std(tracks['r2'])}, Std MAPE: {np.std(tracks['mape'])}")
```
