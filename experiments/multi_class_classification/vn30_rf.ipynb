```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parents[1]
sys.path.append(str(PROJECT_ROOT))
```

```python
import warnings
warnings.filterwarnings("ignore")
from loaders._load_vn30_multi_class import preprocess, VN30, TARGETS
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
X_train, Y_train = preprocess("ACB", verbose=True)["train"]
```

```python
mean_accuracy = 0
for symbol in VN30:
    data = preprocess(symbol, lag=30)

    X_train, Y_train = data['train']
    X_test, Y_test = data['test']
    classes = data['classes']

    model = RandomForestClassifier(
        n_estimators=10,
        max_depth=5,
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, Y_train)
    Y_pred = model.predict(X_test)
    accuracy = accuracy_score(Y_test, Y_pred)
    mean_accuracy += accuracy
    print(f"Accuracy for {symbol}: {accuracy:.4f}")

print(f"Mean accuracy: {mean_accuracy / len(VN30):.4f}")
```
